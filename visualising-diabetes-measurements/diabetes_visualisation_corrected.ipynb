# Diabetes outcome visualisation — corrected portfolio notebook

A separately developed repair of the original submission. Input: `diabetes(1).csv` (768 rows and nine columns), recovered from another task in the original HyperionDev repository. The original notebook is preserved unchanged.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("diabetes(1).csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
# Zero-coded measurements are handled as missing for this exploratory analysis.
columns_to_fix = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]
zero_counts = (df[columns_to_fix] == 0).sum()
print("Zero-coded records before imputation:\n", zero_counts)

for col in columns_to_fix:
    measured = df[col].replace(0, np.nan)
    df[col] = measured.fillna(measured.median())
```

```python
sns.countplot(x="Outcome", data=df)

plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")

plt.show()
```

```python
plt.hist(df["Age"], bins=20)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.boxplot(x="Outcome", y="Glucose", data=df)

plt.title("Glucose Level vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Glucose Level")

plt.show()
```

```python
sns.boxplot(x="Outcome", y="BMI", data=df)

plt.title("BMI vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("BMI")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
sns.scatterplot(x="Age", y="Glucose", hue="Outcome", data=df)

plt.title("Age vs Glucose Level by Diabetes Outcome")

plt.show()
```

## Interpretation and limitations

The supplied dataset contains 768 observations, with 268 records labelled `Outcome = 1` and 500 labelled `Outcome = 0`. These are sample labels, **not** population prevalence estimates.

Glucose, BMI and age differ descriptively across the recorded outcome categories, but the notebook does not test statistical significance or estimate adjusted or causal effects. Its correlation heatmap is exploratory and does not establish predictor importance in a fitted model.

Five measurement columns contain zeros that the original assignment treats as potentially missing. This corrected version replaces those zeros with the median of *nonzero* entries, not the median including zero placeholders. The imputation is a simple educational assumption, not a clinically validated procedure. See the accompanying research paper for sensitivity and provenance limitations.
