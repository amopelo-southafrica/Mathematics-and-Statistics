```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("diabetes(1).csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
columns_to_fix = ["Glucose", "BloodPressure", "SkinThickness", "Insulin", "BMI"]

for col in columns_to_fix:
    df[col] = df[col].replace(0, df[col].median())
```

```python
sns.countplot(x="Outcome", data=df)

plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.ylabel("Number of Patients")

plt.show()
```

```python
plt.hist(df["Age"], bins=20)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
sns.boxplot(x="Outcome", y="Glucose", data=df)

plt.title("Glucose Level vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("Glucose Level")

plt.show()
```

```python
sns.boxplot(x="Outcome", y="BMI", data=df)

plt.title("BMI vs Diabetes Outcome")
plt.xlabel("Outcome")
plt.ylabel("BMI")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```

```python
sns.scatterplot(x="Age", y="Glucose", hue="Outcome", data=df)

plt.title("Age vs Glucose Level by Diabetes Outcome")

plt.show()
```

```python
Findings

Patients with higher glucose levels are significantly more likely to have diabetes.

Individuals with higher BMI values also show increased diabetes prevalence.

Age appears to influence diabetes risk, with older patients showing a higher occurrence of diabetes.

The correlation heatmap indicates glucose is one of the strongest predictors of diabetes.
```
