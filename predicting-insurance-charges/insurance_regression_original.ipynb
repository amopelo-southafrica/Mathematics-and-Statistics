# Insurance Cost Prediction using Linear Regression

## Introduction
This notebook explores the relationship between a person's age and their insurance charges using simple linear regression.

Linear regression is a statistical technique used to model the relationship between a dependent variable and one or more independent variables. In this case, the goal is to determine whether age influences insurance costs.

The analysis will include:
- Visualising the relationship between age and insurance charges
- Training a linear regression model
- Plotting the best-fit regression line
- Interpreting the results

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

## Importing Required Libraries

The following libraries are used in this analysis:

- **pandas** for loading and manipulating the dataset
- **matplotlib** and **seaborn** for data visualisation
- **LinearRegression from sklearn** to build the regression model

These libraries allow us to explore the data and create a predictive model.

```python
df = pd.read_csv("insurance.csv")
df.head()
```

## Loading the Dataset

The dataset used in this analysis is the **insurance dataset**, which contains information about individuals and their medical insurance costs.

The dataset is loaded into a pandas DataFrame so that it can be easily analysed and manipulated.

```python
# Dataset information
df.info()

# Summary statistics
df.describe()

# Check missing values
df.isnull().sum()

# Remove duplicate rows if any
df = df.drop_duplicates()
```

## Visualising the Relationship Between Age and Insurance Charges

A scatter plot is used to visualise the relationship between age and insurance charges.

In this plot:
- The **x-axis represents age**
- The **y-axis represents insurance charges**

This visualisation helps determine whether there is a visible trend between these two variables. If insurance charges increase as age increases, it suggests that age may be a significant predictor of insurance costs.

```python
# scatter plot
sns.scatterplot(x="age", y="charges", data=df)
plt.title("Age vs Insurance Charges")
plt.show()
```

## Building the Linear Regression Model

To quantify the relationship between age and insurance charges, a **linear regression model** is created using the `LinearRegression` class from scikit-learn.

In this model:
- **Age** is the independent variable (X)
- **Insurance charges** are the dependent variable (y)

The model attempts to find the best-fitting straight line that minimises the difference between the predicted values and the observed values in the dataset.

```python
# regression model
X = df[["age"]]
y = df["charges"]
model = LinearRegression()
model.fit(X,y)

y_pred = model.predict(X)
```

## Generating Predictions

Once the regression model has been trained, it can be used to generate predictions.

The model predicts insurance charges based on the age values in the dataset. These predicted values represent the model’s estimate of insurance costs for each age.

```python
# plot regression line
plt.scatter(X,y)
plt.plot(X,y_pred,color="red")
plt.xlabel("Age")
plt.ylabel("Insurance Charges")
plt.title("Linear Regression Fit")
plt.show()
```
